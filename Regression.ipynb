```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

Regression in machine learning is a supervised learning technique used to predict continuous numerical values based on input features. It establishes relationships between independent variables (features) and a dependent variable (target) to make predictions. For example, predicting house prices based on features like size, location, and number of rooms.

<h3>lmplot()</h3>
regression plot is used in machine learning so we will not fully discuss what, how and where it is used but we will see how it is created and what are the values we can put inside it

```python
tips = sns.load_dataset("tips")
```

```python
sns.lmplot(x= 'total_bill', y= 'tip', data=tips)
#Linear Regression: Assumes a linear relationship between features and the target. It can be simple (one feature) or multiple (several features). For example, predicting fuel efficiency based on car weight.
```

```python
sns.lmplot(x= 'total_bill', y= 'tip', data=tips, hue = 'sex')
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette='coolwarm')
```

```python
sns.lmplot(x = 'total_bill',y = 'tip',data = tips,hue = 'sex',palette='coolwarm',markers=['o','v'],
           scatter_kws={'s':50})
```
